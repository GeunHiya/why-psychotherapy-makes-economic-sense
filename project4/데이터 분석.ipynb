```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import statsmodels.api as sm

# =======================================
# 1. 시군구 단위 데이터 불러오기
# =======================================
file_path_si = r"C:\Users\rmsgu\Desktop\project4\processed_data\data_시군구.csv"
# 인코딩 시도: cp949 → utf-8 → euc-kr
try:
    df_si = pd.read_csv(file_path_si, encoding="cp949")
except UnicodeDecodeError:
    try:
        df_si = pd.read_csv(file_path_si, encoding="utf-8")
    except UnicodeDecodeError:
        df_si = pd.read_csv(file_path_si, encoding="euc-kr")

# =======================================
# 2. 시도 단위 데이터 불러오기
# =======================================
file_path_do = r"C:\Users\rmsgu\Desktop\project4\processed_data\data_시도.csv"
try:
    df_do = pd.read_csv(file_path_do, encoding="cp949")
except UnicodeDecodeError:
    try:
        df_do = pd.read_csv(file_path_do, encoding="utf-8")
    except UnicodeDecodeError:
        df_do = pd.read_csv(file_path_do, encoding="euc-kr")

# =======================================
# 3. 시군구 변수 및 데이터 전처리
# =======================================
# 사용할 시군구 변수 5개 + 종속변수
si_features = [
    "인구10만명당_1인세대수",
    "인구10만명당_공원수",
    "인구10만명당_범죄횟수",      # 범죄율
    "진학률",
    "평균소득월액"
]
si_target = "인구_10만명_환자수"
# 필요한 컬럼만 추출하고 결측치 제거
df_si = df_si[si_features + [si_target]].dropna().reset_index(drop=True)

# =======================================
# 4. 시도 변수 및 데이터 전처리
# =======================================
# 사용할 시도 변수: 인구_10만명당_고독사 + 종속변수 컬럼명
do_feature = "인구_10만명당_고독사"
# 시도 데이터에서 실제 컬럼명을 확인한 결과 '인구10만명당_환자수'로 되어 있음
do_target = "인구10만명당_환자수"
df_do = df_do[[do_feature, do_target]].dropna().reset_index(drop=True)

# =======================================
# 5. 회귀 학습: 시군구 5개 변수 (다중회귀)
# =======================================
X_si_const = sm.add_constant(df_si[si_features])
y_si = df_si[si_target]
model_si = sm.OLS(y_si, X_si_const).fit()

# (원한다면 회귀 결과 출력)
coef_table_si = pd.DataFrame({
    "변수": model_si.params.index,
    "계수": model_si.params.values,
    "Std Err": model_si.bse,
    "t-value": model_si.tvalues,
    "p-value": model_si.pvalues
}).round(4)
print("=== 시군구 다중회귀 결과 ===")
print(coef_table_si.to_string(index=False))


# =======================================
# 6. 회귀 학습: 시도 1개 변수 (단순회귀)
# =======================================
X_do = df_do[[do_feature]].values  # (n,1) 형태
y_do = df_do[do_target].values
lr_do = LinearRegression().fit(X_do, y_do)
slope_do = lr_do.coef_[0]
intercept_do = lr_do.intercept_

# =======================================
# 7. 산점도+회귀선 6개 그리기
# =======================================
# 2행 3열 그리드 생성
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
axes = axes.flatten()

# --- 7-1) 시군구 변수 5개 순회 ---
for i, feature in enumerate(si_features):
    ax = axes[i]
    x_vals = df_si[[feature]].values
    y_vals = df_si[si_target].values

    # 단순 회귀(시각화용)
    lr_uni = LinearRegression().fit(x_vals, y_vals)
    slope = lr_uni.coef_[0]
    intercept = lr_uni.intercept_

    ax.scatter(x_vals, y_vals, alpha=0.6, label="Data points")
    x_range = np.linspace(x_vals.min(), x_vals.max(), 100).reshape(-1, 1)
    y_pred = lr_uni.predict(x_range)
    ax.plot(x_range, y_pred, color='red', linewidth=2,
            label=f"y = {slope:.2f}·x {'+' if intercept>=0 else '-'} {abs(intercept):.1f}")

    ax.set_xlabel(feature, fontsize=12)
    ax.set_ylabel(si_target, fontsize=12)
    ax.set_title(f"{feature} vs {si_target}", fontsize=14)
    ax.legend(frameon=True, edgecolor='black')
    ax.grid(True, linestyle='--', alpha=0.5)
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_linewidth(1)

# --- 7-2) 시도 변수 (인구_10만명당_고독사) ---
ax6 = axes[5]
ax6.scatter(X_do, y_do, alpha=0.6, label="Data points")
x_do_range = np.linspace(X_do.min(), X_do.max(), 100).reshape(-1, 1)
y_do_pred = lr_do.predict(x_do_range)
ax6.plot(x_do_range, y_do_pred, color='red', linewidth=2,
         label=f"y = {slope_do:.2f}·x {'+' if intercept_do>=0 else '-'} {abs(intercept_do):.1f}")

ax6.set_xlabel(do_feature, fontsize=12)
ax6.set_ylabel(do_target, fontsize=12)
ax6.set_title(f"{do_feature} vs {do_target}", fontsize=14)
ax6.legend(frameon=True, edgecolor='black')
ax6.grid(True, linestyle='--', alpha=0.5)
for spine in ax6.spines.values():
    spine.set_visible(True)
    spine.set_linewidth(1)

plt.tight_layout()

# =======================================
# 8. 이미지 저장 (바탕화면 그림 폴더)
# =======================================
desktop = os.path.join(os.path.expanduser("~"), "Desktop")
save_dir = os.path.join(desktop, "그림")
os.makedirs(save_dir, exist_ok=True)
plt.savefig(os.path.join(save_dir, "산점도_회귀선_6개.png"), dpi=300)

# =======================================
# 9. 그래프 출력
# =======================================
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import statsmodels.api as sm
import os

# =======================================
# 1. 데이터 불러오기 (시도 단위)
# =======================================
file_path_state = r"C:\Users\rmsgu\Desktop\project4\processed_data\data_시도.csv"
# 인코딩은 'utf-8' 또는 'cp949' 중 하나를 시도하세요.
df_state = pd.read_csv(file_path_state, encoding="utf-8")

# =======================================
# 2. 사용할 변수 목록 (시도 단위)
#    '인구_10만명당_고독사' 제거
# =======================================
features = [
    "1인당 정신건강예산",
    "인구_10만명당_상담건수",
    "인구_10만명당_기관수",
    "상근인력대비 전문인력 비율",
    "인구 10만명당 상근인력",
    "인구 10만명당 전문인력"
]
# 종속변수(환자 수)
dependent = "인구10만명당_환자수"

# =======================================
# 3. 결측치 제거 및 데이터 추출
# =======================================
cols = features + [dependent]
df_state = df_state[cols].dropna()

# =======================================
# 4. 독립변수(X)와 종속변수(y) 설정
# =======================================
y_state = df_state[dependent]
X_state = df_state[features]

# =======================================
# 5. OLS 회귀 분석 (Statsmodels 사용)
# =======================================
X_state_const = sm.add_constant(X_state)           # 상수항 추가
model_state = sm.OLS(y_state, X_state_const).fit()  # 회귀모델 적합

# =======================================
# 6. 회귀 계수와 p-value 요약 (DataFrame)
# =======================================
coef_state_df = pd.DataFrame({
    "변수": model_state.params.index,
    "계수(Coefficient)": model_state.params.values,
    "표준오차(Std Err)": model_state.bse,
    "t-value": model_state.tvalues,
    "p-value": model_state.pvalues
})
coef_state_df = coef_state_df.round(4)
print("=== Regression Coefficients Summary (시도 단위) ===")
print(coef_state_df.to_string(index=False))


# =======================================
# 7. 독립변수별 산점도 및 단순회귀선 그리기 (시도 단위)
#    인구_10만명당_고독사 제외 → 총 6개 변수
# =======================================
n_feats = len(features)  # 이제 6
# 그래프를 2행 3열로 구성: 6개 변수
fig, axes_state = plt.subplots(2, 3, figsize=(18, 12))
axes_state = axes_state.flatten()

for i, feat in enumerate(features):
    ax = axes_state[i]
    x_vals = df_state[[feat]].values        # (n,1) 형태
    y_vals_state = df_state[dependent].values  # (n,) 형태

    # 단순 선형회귀 (Univariate)
    lr_state = LinearRegression()
    lr_state.fit(x_vals, y_vals_state)
    slope = lr_state.coef_[0]
    intercept = lr_state.intercept_

    # 산점도
    ax.scatter(x_vals, y_vals_state, alpha=0.7, label="Data points")

    # 회귀선
    x_range = np.linspace(x_vals.min(), x_vals.max(), 100).reshape(-1, 1)
    y_pred_line = lr_state.predict(x_range)
    ax.plot(
        x_range, y_pred_line, color='red', linewidth=2,
        label=f"y = {slope:.2f}·x {'+' if intercept>=0 else '-'} {abs(intercept):.1f}"
    )

    ax.set_xlabel(feat, fontsize=11)
    ax.set_ylabel(dependent, fontsize=11)
    ax.set_title(f"{feat} vs {dependent}", fontsize=12)
    ax.legend(fontsize=9)
    ax.grid(True, linestyle='--', alpha=0.5)
    for spine in ax.spines.values():
        spine.set_visible(True)
        spine.set_linewidth(1)

# 빈 칸 없으므로 숨길 서브플롯 없음

plt.tight_layout()

# =======================================
# 8. 이미지 저장 (바탕화면 "그림" 폴더)
# =======================================
desktop = os.path.join(os.path.expanduser("~"), "Desktop")
save_dir = os.path.join(desktop, "그림")
os.makedirs(save_dir, exist_ok=True)
plt.savefig(os.path.join(save_dir, "시도_산점도_회귀선_6개.png"), dpi=300)

# =======================================
# 9. 그래프를 화면에 표시
# =======================================
plt.show()
```
